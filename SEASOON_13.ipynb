````python
'''<<------------------------------------------------------------------SEASOON_13---------------------------------------------------------------->>'''
'''1.Write a recursive function in Python called print_playlist_songs(songs) that takes a list of song names (like a Spotify playlist) and prints each song name 
one by one using recursion.'''
def print_play_list_songs(songs):
    #base case
    if len(songs)==0:
        return
    print(songs[0])
    #Recursive case 
    print_play_list_songs(songs[1:])

play_list = [    "Shape of You",
    "Blinding Lights",
    "Levitating",
    "Senorita",
    "Perfect"
]
print_play_list_songs(play_list)
````

````python
'''2.Create a recursive function count_unread_messages(messages) that takes a nested dictionary representing WhatsApp chat groups and subgroups, and returns 
the total number of unread messages across all groups.<br><br><em><strong>Hint:</strong> Each group can have a 'count' key for unread messages and a 'subgroups' 
key with a list of more groups.</em>'''
def unread_messages(messages):
    total = messages["count"]

    for group in messages["subgroups"]:
        total = total + unread_messages(group)

    return total


groups = {
    "name": "WhatsApp",
    "count": 5,
    "subgroups": [
        {
            "name": "Friends",
            "count": 3,
            "subgroups": []
        },
        {
            "name": "Family",
            "count": 2,
            "subgroups": [
                {
                    "name": "Cousins",
                    "count": 4,
                    "subgroups": []
                }
            ]
        }
    ]
}

print("Total unread messages =", unread_messages(groups))
````

````python
'''4.Build a recursive function format_number_short(n) that takes a number (like a follower count on Instagram or YouTube) and returns it as a string in 
short format: 1500 as '1.5K', 1200000 as '1.2M', 500 as '500'.'''
def number_short(n, level=0):
    if n < 1000:
        if level == 0:
            return str(int(n))
        elif level == 1:
            return str(round(n, 1)) + "K"
        else:
            return str(round(n, 1)) + "M"

    return number_short(n / 1000, level + 1)


# Test cases
print(number_short(500))
print(number_short(1500))
print(number_short(1200000))
print(number_short(2500000))
````
