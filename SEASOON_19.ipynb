```python
'''3.Create a class called FoodOrder with attributes: restaurant_name, items (list), and total_price. Write an __init__() constructor to initialize these, then create an object representing your last Zomato or Swiggy order and print its details.'''

class FoodOrder:
    def __init__(self, restaurant_name, items, total_price):
        self.restaurant_name = restaurant_name
        self.items = items
        self.total_price = total_price


order = FoodOrder(
    "Domino's Pizza",
    ["pizza",
     "burgwe", "Coke"],
    789
    
)

print("Restaurant Name:", order.restaurant_name)
print("Items:", order.items)
print("Total Price: ₹", order.total_price)
```

```python
'''4.Extend the FoodOrder class by adding a method add_item(self, item_name, item_price) that adds the item to the items list and updates total_price. Demonstrate by adding two items to your order and printing the updated total.'''
class FoodOrder:
    def __init__(self,resturant_name,total_price,items):
        self.resturant_name = resturant_name
        self.total_price = total_price
        self.items = items
    def add_items(self,items_name,item_price):
        self.items.append(items_name)
        self.total_price+=item_price

order = FoodOrder("The Honest",1200,["pulav","pizza","idli sambhar","utapam","tomato soup"])
order.add_items("dal rice",240)
order.add_items("gulab jamun",190)

print("Resturant Name :--",order.resturant_name)
print("Total_Price :--",order.total_price)
print("Items :--",order.items)
```
